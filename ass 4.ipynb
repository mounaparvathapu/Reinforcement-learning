```python
import math

# ============================================================
# Assignment: Upper Confidence Bound (UCB)
# Multi-Armed Bandit Problem
# ============================================================

# ------------------------------------------------------------
# Part A - Machine Initialization
# ------------------------------------------------------------

machines = ['A', 'B', 'C']

M = 3
N_play = 10

# ------------------------------------------------------------
# Part B - Initialize Knowledge Tables
# ------------------------------------------------------------

N = {'A': 0, 'B': 0, 'C': 0}        # Participation count
R = {'A': 0, 'B': 0, 'C': 0}        # Total reward
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}  # Estimated value
UCB = {'A': 0.0, 'B': 0.0, 'C': 0.0} # UCB value

# Given reward patterns
reward_pattern = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

# Index to keep track of rewards
reward_index = {
    'A': 0,
    'B': 0,
    'C': 0
}

total_reward = 0

print("==============================================")
print("      UPPER CONFIDENCE BOUND (UCB)")
print("==============================================")
print("Machines:", machines)
print("Number of plays:", N_play)
print()

# ------------------------------------------------------------
# Part C - UCB Calculation and Machine Selection
# ------------------------------------------------------------

for t in range(1, N_play + 1):

    # Play each machine once initially
    if t <= M:

        selected_machine = machines[t - 1]

    else:

        # Calculate UCB for each machine
        for machine in machines:

            UCB[machine] = Q[machine] + math.sqrt(
                (2 * math.log(t)) / N[machine]
            )

        # Select machine with highest UCB
        selected_machine = max(
            machines,
            key=lambda machine: UCB[machine]
        )

    # --------------------------------------------------------
    # Part D - Reward Collection
    # --------------------------------------------------------

    # Use rewards in repeating order
    index = reward_index[selected_machine]

    reward = reward_pattern[selected_machine][
        index % len(reward_pattern[selected_machine])
    ]

    reward_index[selected_machine] += 1

    # --------------------------------------------------------
    # Part E - Q-Value Update
    # --------------------------------------------------------

    # Update participation count
    N[selected_machine] += 1

    # Update total reward
    R[selected_machine] += reward

    # Incremental sample-average formula:
    # Qnew = Qold + (R - Qold) / N

    Q[selected_machine] = Q[selected_machine] + (
        reward - Q[selected_machine]
    ) / N[selected_machine]

    # Add to total reward
    total_reward += reward

    # Calculate UCB values
    for machine in machines:

        if N[machine] == 0:
            UCB[machine] = float('inf')

        else:
            UCB[machine] = Q[machine] + math.sqrt(
                (2 * math.log(t)) / N[machine]
            )

    # --------------------------------------------------------
    # Display information for each play
    # --------------------------------------------------------

    print("----------------------------------------------")
    print("Play:", t)
    print("Selected Machine:", selected_machine)
    print("Reward:", reward)

    print("N:", N)
    print("R:", R)

    print("UCB:", {
        machine:
        "inf" if UCB[machine] == float('inf')
        else round(UCB[machine], 3)
        for machine in machines
    })

    print("Q:", {
        machine: round(Q[machine], 3)
        for machine in machines
    })


# ============================================================
# Part F - Final Results
# ============================================================

print("\n==============================================")
print("                FINAL RESULTS")
print("==============================================")

# Calculate final UCB
for machine in machines:

    UCB[machine] = Q[machine] + math.sqrt(
        (2 * math.log(N_play)) / N[machine]
    )

print("\nMachine\tN\tR\tUCB\t\tQ")
print("----------------------------------------------")

for machine in machines:

    print(
        f"{machine}\t"
        f"{N[machine]}\t"
        f"{R[machine]}\t"
        f"{UCB[machine]:.3f}\t\t"
        f"{Q[machine]:.3f}"
    )

# ------------------------------------------------------------
# Total Reward
# ------------------------------------------------------------

print("\nTotal Reward Received by Player:", total_reward)

# ------------------------------------------------------------
# Best Machine
# ------------------------------------------------------------

best_machine = max(machines, key=lambda machine: Q[machine])

print("Best Machine based on Q-value:", best_machine)

print("\n==============================================")
print("             PROGRAM COMPLETED")
print("==============================================")
```
