```python
# Greedy Algorithm for Multi-Armed Bandit
# Using Arithmetic Mean Method

# Step 1: Take the number of machines
n = int(input("Enter the number of machines: "))

machines = {}

# Step 2: Take rewards for each machine
for i in range(n):
    name = input(f"Enter the name of machine {i + 1}: ")

    rewards = list(map(int, input(
        f"Enter the observed rewards for Machine {name} separated by spaces: "
    ).split()))

    machines[name] = rewards

# Step 3: Calculate Q(a) for each machine
q_values = {}

for name, rewards in machines.items():
    q = sum(rewards) / len(rewards)
    q_values[name] = q

# Step 4: Display the estimated Q(a)
print("\n----- Estimated Values -----")

for name, rewards in machines.items():
    print(f"Machine {name}")
    print(f"Rewards: {rewards}")
    print(f"Q({name}) = {q_values[name]:.2f}")
    print()

# Step 5: Greedy selection
selected_machine = max(q_values, key=q_values.get)

print(f"Greedy Selected Machine: {selected_machine}")
```

```python
# Greedy Algorithm for Multi-Armed Bandit
# Incremental Update Method

machines = {
    "A": [3, 2, 4],
    "B": [6, 7],
    "C": [5, 4, 6]
}

# Initialize Q and N
Q = {"A": 0, "B": 0, "C": 0}
N = {"A": 0, "B": 0, "C": 0}

print("Initial Values:")
print("Q(A) = 0, Q(B) = 0, Q(C) = 0")
print("N(A) = 0, N(B) = 0, N(C) = 0")

# Update after every play
for machine, rewards in machines.items():

    for reward in rewards:

        # Step 1: Increment N(a)
        N[machine] += 1

        # Step 2: Update Q(a)
        Q[machine] = Q[machine] + (reward - Q[machine]) / N[machine]

        # Step 3: Find current highest Q
        selected_machine = max(Q, key=Q.get)

        print("\n----------------------------")
        print(f"Machine Played: {machine}")
        print(f"Reward: {reward}")

        print(f"Q(A) = {Q['A']:.2f}")
        print(f"Q(B) = {Q['B']:.2f}")
        print(f"Q(C) = {Q['C']:.2f}")

        print(f"Current Greedy Machine: {selected_machine}")

# Final result
print("\n============================")
print("Final Q Values")
print("============================")

for machine in machines:
    print(f"Q({machine}) = {Q[machine]:.2f}")

print(f"\nGreedy Selected Machine: {max(Q, key=Q.get)}")
```
