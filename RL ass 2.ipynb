```python
import random

# ============================================================
# PART A – Machine Initialization
# ============================================================

machines = ['A', 'B', 'C']

# Number of machines
M = 3

# Number of plays
N_play = 10

# Exploration probability
epsilon = 0.2

# ============================================================
# PART B – Initialize N, R and Q tables
# ============================================================

# N = number of times each machine has been selected
N = {machine: 0 for machine in machines}

# R = total reward received from each machine
R = {machine: 0 for machine in machines}

# Q = estimated value of each machine
Q = {machine: 0.0 for machine in machines}

# Total reward received by player
total_reward = 0

print("Machines:", machines)
print("Number of plays:", N_play)
print("Epsilon:", epsilon)
print("\nInitial Values:")
print("N =", N)
print("R =", R)
print("Q =", Q)

# ============================================================
# PART C – ε-Greedy Decision
# ============================================================

for play in range(1, N_play + 1):

    # Generate random number between 0 and 1
    r = random.random()

    print("\n" + "=" * 45)
    print("Play", play)
    print("Random number r =", round(r, 4))

    # Exploration
    if r < epsilon:
        selected_machine = random.choice(machines)
        decision = "Exploration"

    # Exploitation
    else:
        max_Q = max(Q.values())

        # If multiple machines have the same highest Q,
        # randomly choose among them
        best_machines = [
            machine for machine in machines
            if Q[machine] == max_Q
        ]

        selected_machine = random.choice(best_machines)
        decision = "Exploitation"

    print("Decision:", decision)
    print("Selected Machine:", selected_machine)

    # ========================================================
    # PART D – Reward Collection
    # ========================================================

    reward = float(input(
        f"Enter reward obtained from Machine {selected_machine}: "
    ))

    # ========================================================
    # PART E – Q-Value Update
    # ========================================================

    # Increase number of times selected
    N[selected_machine] += 1

    # Add reward to total reward of that machine
    R[selected_machine] += reward

    # Incremental sample-average formula:
    #
    # Q_new = Q_old + (Reward - Q_old) / N
    #
    old_Q = Q[selected_machine]

    Q[selected_machine] = (
        old_Q +
        (reward - old_Q) / N[selected_machine]
    )

    # Add to player's total reward
    total_reward += reward

    print("Updated N:", N[selected_machine])
    print("Updated R:", R[selected_machine])
    print("Updated Q:", round(Q[selected_machine], 4))

# ============================================================
# PART F – Display Final Results
# ============================================================

print("\n\n")
print("=" * 60)
print("FINAL RESULTS")
print("=" * 60)

print("\nMachine\tN (Count)\tR (Total Reward)\tQ (Estimated Value)")
print("-" * 60)

for machine in machines:
    print(
        f"{machine}\t{N[machine]}\t\t"
        f"{R[machine]:.2f}\t\t\t"
        f"{Q[machine]:.4f}"
    )

print("\nTotal Reward received after 10 plays =", total_reward)

# Best machine according to final Q-value
best_machine = max(Q, key=Q.get)

print("Best Machine according to final Q-value:", best_machine)
```
