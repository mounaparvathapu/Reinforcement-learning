```python
# ============================================================
# Assignment: Building a Random Agent for Reinforcement
# Learning Environments
# ============================================================

# Install Gymnasium
!pip install -q gymnasium[classic-control,toy-text]

import gymnasium as gym
import random


# ============================================================
# Random Agent Function
# ============================================================

def random_agent(env_name):

    print("\n==========================================")
    print("Environment:", env_name)
    print("==========================================")

    # Part 1: Create Environment
    env = gym.make(env_name)

    # Part 2: Reset Environment
    state, info = env.reset()

    print("Initial State:", state)

    # Part 3: Identify Action Space
    print("Action Space:", env.action_space)

    total_reward = 0
    steps = 0

    # Part 4: Continue until episode terminates
    while True:

        # Part 5: Select random action
        action = env.action_space.sample()

        # Part 6: Execute action
        next_state, reward, terminated, truncated, info = env.step(action)

        # Part 7: Observe next state and reward
        steps += 1
        total_reward += reward

        # Update state
        state = next_state

        # Part 8: Check whether episode has terminated
        if terminated or truncated:
            break

    # Part 9: Calculate average reward
    average_reward = total_reward / steps

    print("Number of Steps:", steps)
    print("Total Reward:", total_reward)
    print("Average Reward:", round(average_reward, 3))

    # Close environment
    env.close()

    return total_reward, average_reward


# ============================================================
# 1. CartPole
# ============================================================

cartpole_reward, cartpole_average = random_agent(
    "CartPole-v1"
)


# ============================================================
# 2. MountainCar
# ============================================================

mountaincar_reward, mountaincar_average = random_agent(
    "MountainCar-v0"
)


# ============================================================
# 3. FrozenLake
# ============================================================

frozenlake_reward, frozenlake_average = random_agent(
    "FrozenLake-v1"
)


# ============================================================
# Final Summary
# ============================================================

print("\n\n==========================================")
print("              FINAL SUMMARY")
print("==========================================")

print("\nEnvironment\t\tTotal Reward\tAverage Reward")
print("-----------------------------------------------")

print(
    f"CartPole-v1\t\t{cartpole_reward:.2f}\t\t{cartpole_average:.3f}"
)

print(
    f"MountainCar-v0\t\t{mountaincar_reward:.2f}\t\t{mountaincar_average:.3f}"
)

print(
    f"FrozenLake-v1\t\t{frozenlake_reward:.2f}\t\t{frozenlake_average:.3f}"
)
```
